```python
import numpy as np
import pygetm

%matplotlib widget
```

```python
# Cartesian grid defined in terms of coordinates of centres (T points)
# Alternatively it could be defined in terms of coordinates of corners (X points)
# To do so, add interfaces=True in the call to create_cartesian
x = np.linspace(0.0, 1000.0, 100)
y = np.linspace(0.0, 500.0, 50)
dom = pygetm.domain.create_cartesian(x=x, y=y, f=0.0, H=1.0)
dom.plot(show_mesh=True)
```

```python
# Mock curvilinear grid, with x linearly increasing, and y a hyberbolic function of x (not great for orthogonality!)
# It is defined in terms of number of tracer points (nx, ny) and the 2D coordinates of the corner (X) points (x, y)
nx, ny = 100, 50
x = np.linspace(0.0, 1000.0, nx + 1)[np.newaxis, :]
y_extent = 500 * (0.8 - x / (x + 500.0))
y = np.linspace(-1.0, 1.0, ny + 1)[:, np.newaxis] * y_extent
dom = pygetm.domain.create_cartesian(x, y, f=0.0, H=1.0, interfaces=True)
dom.plot(show_mesh=True)
```
